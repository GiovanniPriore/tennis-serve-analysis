# Chronological model evaluation

Model selection uses 2021–2022 validation MAE. The selected pipeline is evaluated once on 5,698 matches from 2023–2024. This is post-match explanatory regression, not pre-match forecasting.

```python
from pathlib import Path
import json, pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
validation = pd.read_csv(ROOT / 'artifacts/validation_results.csv')
metrics = json.loads((ROOT / 'reports/metrics.json').read_text())
display(validation.head(12), pd.Series(metrics, name='test result'))
```

```python
from IPython.display import Image, display
for name in ['predicted_vs_actual.png','residual_distribution.png','permutation_importance.png']:
    display(Image(filename=ROOT / 'reports/figures' / name))
```

## Main result

Combined-feature XGBoost reaches test MAE 1.084 games (95% bootstrap CI 1.064–1.105), RMSE 1.368 and R² 0.713. Importance is associative and partly reflects the mechanical connection between in-match point statistics and final games.
